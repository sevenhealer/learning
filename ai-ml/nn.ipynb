```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# to get the same sequence of ramdom numbers every time
np.random.seed(42)
```

```python
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([
    [0],
    [1],
    [1],
    [0]
])
```

```python
print("XOR Dataset")
print("-"*30)
for i in range(len(X)):
    print(f"Input: {X[i]} -> Output: {y[i]}")
```

```python
plt.figure(figsize = (8, 6))

for i in range(len(X)):
    color = 'red' if y[i][0]==0 else 'blue'
    marker = 'o' if y[i][0]==0 else 's'
    plt.scatter(X[i][0], X[i][1], c=color, s=200, marker=marker)

plt.xlabel('Input A', fontsize=12)
plt.ylabel('Input B', fontsize=12)
plt.title('XOR Problem: Can not seperate red and blue with single straight line')
plt.xlim(-0.5, 1.5)
plt.ylim(-0.5, 1.5)
plt.legend(['Class 0', 'Class 1'], loc='upper right')

plt.show()
```
